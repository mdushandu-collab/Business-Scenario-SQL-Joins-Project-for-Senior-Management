```sql
SELECT (*)
FROM workspace.default.bright_tv_dataset
LIMIT 100 ;

SELECT (*)
FROM workspace.default.bright_tv_viewership_dataset
LIMIT 100 ;

SELECT COUNT(*)
FROM workspace.default.bright_tv_dataset
LIMIT 100 ;

SELECT COUNT (*)
FROM workspace.default.bright_tv_viewership_dataset ;

----------------------------------------------------------------------
--**SUBSCRIBER CHECKS**--
----------------------------------------------------------------------

SELECT COUNT (*) AS total_subscribers
FROM workspace.default.bright_tv_dataset  ;

--** THERE IS A TOTAL OF 5375 SUBSCRIBERS**--

SELECT COUNT (*) AS total_platform_activity
FROM workspace.default.bright_tv_viewership_dataset  ;

----------------------------------------------------------------------------
--**GROSS MISSING VALUES ANALYSIS**--
----------------------------------------------------------------------------
SELECT
       SUM(CASE WHEN Name IS NULL THEN 1 ELSE 0 END) AS Missing_Name,
       SUM(CASE WHEN Gender IS NULL THEN 1 ELSE 0 END) AS Missing_Gender,
       SUM(CASE WHEN Race IS NULL THEN 1 ELSE 0 END) AS Missing_Race,
       SUM(CASE WHEN Province IS NULL THEN 1 ELSE 0 END) AS Missing_Province
FROM workspace.default.bright_tv_dataset ;

-----------------------------------------------------------------------
UserID Checks
-----------------------------------------------------------------------
SELECT COUNT(UserID) AS num_of_users,
       COUNT (DISTINCT UserID) AS num_of_distinct_users
FROM workspace.default.bright_tv_dataset ;

SELECT UserID
FROM workspace.default.bright_tv_dataset
WHERE UserID IS NULL ;

SELECT COUNT (*)
FROM workspace.default.bright_tv_dataset

--------------------------------------------------------------
--**Gender Checks**--
--------------------------------------------------------------
SELECT COUNT(DISTINCT Gender) AS num_of_distinct_users,
       COUNT(Gender) AS num_of_users
FROM workspace.default.bright_tv_dataset;

SELECT DISTINCT Gender AS distinct_genders
FROM workspace.default.bright_tv_dataset;

--**GENDER DISTRIBUTION**--
SELECT Gender,
       COUNT (*) AS Customers,
       CASE 
             WHEN Gender = ' ' THEN 'Unknown'
             WHEN Gender = 'None' THEN 'Unknown'
             ELSE Gender
       END AS Clean_gender
FROM workspace.default.bright_tv_dataset
GROUP BY Gender
ORDER BY Customers DESC ;

SELECT Gender,
       CASE 
       WHEN Gender = ' ' THEN 'Unknown'
       WHEN Gender = 'None' THEN 'Unknown'
       ELSE Gender
       END AS Clean_gender
FROM workspace.default.bright_tv_dataset ;

SELECT 
      COUNT (DISTINCT UserID) AS number_of_subcribers,
      CASE 
          WHEN Gender=' ' THEN 'Unknown'
          When Gender='None' THEN 'Unknown'
          ELSE Gender
FROM workspace.default.bright_tv_dataset
GROUP BY Gender

-----------------------------------------------------------------
---**RACE CHECKS**--
------------------------------------------------------------------
SELECT DISTINCT (Race) AS distinct_race
FROM workspace.default.bright_tv_dataset;

SELECT Gender,
       Race,
  CASE 
       WHEN Gender = ' ' THEN 'Unknown'
       WHEN Gender = 'None' THEN 'Unknown'
       ELSE Gender
       END AS Clean_gender,
       CASE 
       WHEN Race = 'other' THEN 'Unknown'
       WHEN Race = 'None' THEN 'Unknown'
       WHEN Race = ' ' THEN 'Unknown'
       ELSE Race 
       END AS Clean_race
FROM workspace.default.bright_tv_dataset ;

-------------------------------------------------------------------------
--**SOCIAL MEDIA AND EMAIL CHECKS**--
--------------------------------------------------------------------------

SELECT COUNT (DISTINCT Email) AS num_of_distinct_emails
FROM workspace.default.bright_tv_dataset  ;

SELECT COUNT (DISTINCT `Social Media Handle`) AS distinct_social_media_handles
FROM workspace.default.bright_tv_dataset ;

SELECT COUNT (`Social Media Handle`) AS num_of_handles
FROM workspace.default.bright_tv_dataset  ;

SELECT UserID,
        Email, 
        `Social Media Handle`,   
       CASE 
            WHEN Email IS NULL THEN '0'
            WHEN Email = ' ' THEN '0'
            WHEN Email = 'None' THEN '0'
            WHEN Email = 'other' THEN '0'
       ELSE Email
       END AS Clean_email,
       CASE 
            WHEN `Social Media Handle` IS NULL THEN '0'
            WHEN `Social Media Handle` = ' ' THEN '0'
            WHEN `Social Media Handle` = 'None' THEN '0'
            WHEN `Social Media Handle` = 'other' THEN '0'
       ELSE `Social Media Handle`
       END AS Clean_social_media_handle
FROM workspace.default.bright_tv_dataset  ;

------------------------------------------------------------------------
--**PROVINCE CHECKS**--
------------------------------------------------------------------------

SELECT DISTINCT (Province) AS distinct_province
FROM workspace.default.bright_tv_dataset  ;

SELECT DISTINCT (Province) AS distinct_province,
       Province,
       CASE 
            WHEN Province = ' ' THEN 'Unknown'
            WHEN Province = 'None' THEN 'Unknown'
            Else Province
       END AS filtered_province
FROM workspace.default.bright_tv_dataset ;

--**PROVINCE DISTRIBUTION**--
SELECT Province,
       COUNT (*) AS Subscribers,
        CASE 
            WHEN Province = ' ' THEN 'Unknown'
            WHEN Province = 'None' THEN 'Unknown'
            ELSE Province
       END AS filtered_province
FROM workspace.default.bright_tv_dataset
GROUP BY Province
ORDER BY Subscribers DESC ;
--**THIS INDICATES THE REGIONS WHICH SHOULD BE TARGETED FOR MARKETING**--

---------------------------------------------------------------------------------------------------
--**AGE CHECKS**--
----------------------------------------------------------------------------------------------------
SELECT
       AVG (Age) AS Avg_age,
       MIN (Age) AS Min_age,
       MAX (Age) AS Max_age
FROM workspace.default.bright_tv_dataset
 ;
 --**Average age**--

 SELECT
       CASE
            WHEN Age < 18 THEN 'Under 18'
            WHEN Age BETWEEN 18 AND 24 THEN '18-24'
            WHEN Age BETWEEN 25 AND 34 THEN '25-34'
            WHEN Age BETWEEN 35 AND 44 THEN '35-44'
            WHEN Age BETWEEN 45 AND 54 THEN '45-54'
       ELSE '55+'
       END AS Age_Group,
       COUNT(*) AS Customers
FROM workspace.default.bright_tv_dataset
GROUP BY Age_Group;

--**Age Bands**--

-------------------------------------------------------------------------------------
--**VIEWERSHIP EXPLORATION**--
-------------------------------------------------------------------------------------
SELECT (*)
FROM workspace.default.bright_tv_viewership_dataset  
LIMIT 100 ;

SELECT
       COUNT (UserID0) AS num_of_users,
       COUNT (DISTINCT UserID0) AS num_of_active_users
FROM workspace.default.bright_tv_viewership_dataset
 ;
--**NOT ALL USERS ARE ACTIVE ON THE PLATFORM**--
--------------------------------------------------------------------------------------------
--**CHANNEL CHECKS**--
--------------------------------------------------------------------------------------------
SELECT
COUNT(DISTINCT Channel2) AS Total_Channels
FROM workspace.default.bright_tv_viewership_dataset ;
--**TOTAL NUMBER OF CHANNELS**--

-------------------------------------------------------------------------------------------
--**CHANNEL CHECKS**--
-------------------------------------------------------------------------------------------
SELECT
       COUNT(DISTINCT Channel2) AS Total_Channels
FROM workspace.default.bright_tv_viewership_dataset  ;
--**TOTAL NUMBER OF CHANNELS**--

SELECT DISTINCT Channel2
FROM workspace.default.bright_tv_viewership_dataset ;

SELECT 
      CASE 
          WHEN Channel2 IN ('SawSee','Sawsee') THEN 'SawSee'
          WHEN Channel2 IN ('SuperSport Live Events','Live on SuperSport','DStv Events 1','Supersport Live Events') THEN 'Live Events'
          ELSE Channel2
      END AS Tv_Channel
FROM workspace.default.bright_tv_viewership_dataset  ;
--**CLEANING DUPLICATES IN CHANNELS**--

SELECT
       Channel2,
       COUNT(*) AS Total_views
FROM workspace.default.bright_tv_viewership_dataset
GROUP BY Channel2
ORDER BY Views DESC;
--**NUMBER OF VIEWS PER CHANNEL IN DESCENDING ORDER**--

SELECT
      Channel2,
      COUNT(*) AS Total_views
FROM workspace.default.bright_tv_viewership_dataset
GROUP BY Channel2
HAVING COUNT(*) < 35 ; 
---**USED TO IDENTIFY UNPOPULAR CHANNELS**--

--------------------------------------------------------------------------------------
--**TIME ANALYSIS**--
--------------------------------------------------------------------------------------
SELECT
      DAYNAME(RecordDate2) AS Days,
      COUNT(*) AS Total_views
FROM workspace.default.bright_tv_viewership_dataset
GROUP BY Days
ORDER BY Total_views DESC  ;
--**TOTAL VIEWS PER DAY**--

SELECT RecordDate2,
       `Duration 2`,
       DAYNAME (RecordDate2) AS Days,
       MONTHNAME (RecordDate2) AS Months,
       DAYNAME (`Duration 2`) AS Duration_days,
       MONTHNAME (`Duration 2`) AS Duration_months
FROM workspace.default.bright_tv_viewership_dataset;

---------------------------------------------------------------------------------------------------
--**USER_ID AND DATE FUNCTIONS & CTE MERGING**--
-----------------------------------------------------------------------------------------------------

WITH Viewership AS (
SELECT --**DATES**
       COALESCE(UserID0, userid4,0) AS user_id,
       TO_CHAR(RecordDate2, 'yyyyMM') AS month_id,
       TO_CHAR(RecordDate2, 'dd') AS watch_day,
       TO_DATE(RecordDate2) AS watch_date,
       DAYNAME(RecordDate2) AS day_of_week,
       
       --**TIME**--
       DATE_FORMAT(RecordDate2, 'HH:mm:ss') AS watch_time,

       CASE
           WHEN DATE_FORMAT(RecordDate2, 'HH:mm:ss') BETWEEN '00:00:00' AND '05:59:00' THEN 'Pre Dawn'
           WHEN DATE_FORMAT(RecordDate2, 'HH:mm:ss') BETWEEN '06:00:00' AND '11:59:00' THEN 'Morning'
           WHEN DATE_FORMAT(RecordDate2, 'HH:mm:ss') BETWEEN '12:00:00' AND '16:59:00' THEN 'Afternoon'
           WHEN DATE_FORMAT(RecordDate2, 'HH:mm:ss') BETWEEN '17:00:00' AND '20:59:00' THEN 'Evening'
           WHEN DATE_FORMAT(RecordDate2, 'HH:mm:ss') BETWEEN '21:00:00' AND '23:59:00' THEN 'Night'
       END AS watch_period,

       HOUR(RecordDate2) AS watch_hour,

       DATE_FORMAT(`Duration 2`, 'HH:mm:ss') AS duration,
       CASE
           WHEN DATE_FORMAT(`Duration 2`, 'HH:mm:ss') BETWEEN '00:00:01' AND '00:30:00' THEN '1.Low_usage'
           WHEN DATE_FORMAT(`Duration 2`, 'HH:mm:ss') BETWEEN '00:30:01' AND '00:59:59' THEN '2.Medium_usage'
           WHEN DATE_FORMAT(`Duration 2`, 'HH:mm:ss') > '00:59:59' THEN '3.High_usage'
       END AS screen_time_buckets,

       CASE
           WHEN DAYNAME(RecordDate2) = 'Sat' OR DAYNAME(RecordDate2) = 'Sun' THEN 'Weekend'
           ELSE 'Weekday'
       END AS day_classification,

       MONTHNAME(RecordDate2) AS month_name,

       Channel2,

       CASE 
          WHEN Channel2 IN ('SawSee','Sawsee') THEN 'SawSee'
          WHEN Channel2 IN ('SuperSport Live Events','Live on SuperSport','DStv Events 1','Supersport Live Events') THEN 'Live Events'
          ELSE Channel2
      END AS Tv_Channel
      
FROM workspace.default.bright_tv_viewership_dataset
),

User_Profiles AS (
       SELECT UserID,
      
        CASE 
            WHEN Province = ' ' THEN 'Unknown'
            WHEN Province = 'None' THEN 'Unknown'
            Else Province
       END AS Region,

       age,
       CASE
            WHEN Age = 0 THEN 'Infants'
            WHEN Age BETWEEN 1 AND 12 THEN 'Kids'
            WHEN Age BETWEEN 13 AND 19 THEN 'Teenager'
            WHEN Age BETWEEN 20 AND 35 THEN 'Youth'
            WHEN Age BETWEEN 36 AND 50 THEN 'Adult'
            WHEN Age BETWEEN 51 AND 65 THEN 'Elder'
            WHEN Age > 65 Then 'Pensioner'
       END AS Age_Group,

       CASE
           WHEN (Email IS NOT NULL) OR (Email <> ' ') OR (Email NOT IN ('None')) THEN 1
           ELSE 0
           END AS email_flag,

       CASE
           WHEN (`Social Media Handle` IS NOT NULL) OR (`Social Media Handle` <> ' ') OR (`Social Media Handle` NOT IN ('None')) THEN 1
           ELSE 0
           END AS sm_flag,

        CASE 
            WHEN Race ILIKE '%other%' THEN 'Unknown'
            WHEN Race ILIKE '%None%' THEN 'Unknown'
            WHEN Race = ' ' THEN 'Unknown'
       ELSE Race 
       END AS Clean_Race,

        CASE 
       WHEN Gender = ' ' THEN 'Unknown'
       WHEN Gender = 'None' THEN 'Unknown'
       ELSE Gender
       END AS Clean_Gender

FROM workspace.default.bright_tv_dataset
)
SELECT COUNT(B.UserID) AS user_id,
       UserID,
       month_id,
       watch_day,
       watch_date,
       day_of_week,
       watch_time,
       watch_period,
       watch_hour,
       duration,
       screen_time_buckets,
       day_classification,
       month_name,
       Region,
       Age,
       Age_Group,
       email_flag,
       sm_flag,
       Clean_Race,
       Clean_Gender,
       Tv_Channel

FROM Viewership A
LEFT JOIN User_Profiles B
ON A.user_id = B.UserID
GROUP BY ALL
;



)


```
